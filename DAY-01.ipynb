```python
#user based non parameters function
def greeting ():
  a=10
  b=20
  print(a+b)
  print ("Hello World")

greeting ()
```

```python
#user based  parameters function
def details (First_Name):
  print (f"My name is {First_Name}")
  print ("my Name is ", First_Name)

details("Vansh")
details("Saloni")
```

```python
temp = float(input("Enter Temperature :"))

if temp >= 30:
  print("Hot")
else:
  print("Cold")
```

```python
#functions with multiple value and return
def add (x,y):
  print("The addition of 2 no is :", x+y)
  print(x-y)
  print(x*y)
  print(x/y)
  return x+y

add(10,30)
```

```python
def greet(name="World"):
    print("Hello", name)

greet()
```

```python
#Keyword Arguments Function
def info(name, age, last_name):
    print(name, age,last_name)

info(24, 24 , "Manch")
info(age=14, name="Vansh" , last_name="Jingar")
```

```python
#multiple arguments functions
def add(*names):
    print((names))

add("A", "B", "C", "D")
add(1,2,3,4,5,6,7,8,9,0)
```

```python
#kwargs functions
def display(**details):
    print(details)

display(name="Vansh", age=21,address="Nandurbar")
display(name="Vansh", age="21",address="Nandurbar", Phone
        =2345678, ht=5.38)
```

```python
# Data Type Identification
value = input("Enter a value: ")

try:
    int(value)
    print("Data Type: int")
except ValueError:
    try:
        float(value)
        print("Data Type: float")
    except ValueError:
        print("Data Type: string")
```
