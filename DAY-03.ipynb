7. Reverse a number using a loop

```python
num = int(input("Enter a number: "))
reverse = 0
while num != 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number:", reverse)
```

8. Check whether a number is prime

```python
num = int(input("Enter a number: "))

if num <= 1:
    print("Not a prime number")
else:
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print("Prime number")
    else:
        print("Not a prime number")
```

9. Function to check if a number is palindrome

```python
def palindrome(num):
    original = num
    reverse = 0

    while num != 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num = num // 10

    if original == reverse:
        return True
    else:
        return False


num = int(input("Enter a number: "))
if palindrome(num):
    print("Palindrome number")
else:
    print("Not a palindrome number")
```

10. Function to calculate sum of digits

```python
def sum_of_digits(num):
    total = 0

    while num != 0:
        digit = num % 10
        total = total + digit
        num = num // 10

    return total


num = int(input("Enter a number: "))
result = sum_of_digits(num)
print("Sum of digits:", result)
```

23. Count frequency of characters in a string

```python
text = "hello"
frequency = {}
for char in text:
    if char in frequency:
        frequency[char] = frequency[char] + 1
    else:
        frequency[char] = 1

print("Character frequency:", frequency)
```

27. Create a class Rectangle to calculate area and perimeter

```python
class Rectangle:

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r = Rectangle(10, 5)
print("Area =", r.area())
print("Perimeter =", r.perimeter())
```

28. Create a class Employee with salary and method to increase salary

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increase_salary(self, amount):
        self.salary = self.salary + amount

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

employee1 = Employee("Ankash", 30000)
print("Before increase:")
employee1.display()
employee1.increase_salary(5000)

print("\nAfter increase:")
employee1.display()
```

29. Create a class ShoppingCart using dictionary and calculate total bill

```python
class ShoppingCart:

    def __init__(self):
        self.items = {}

    def add_item(self, item, price):
        self.items[item] = price

    def total_bill(self):
        total = 0

        for price in self.items.values():
            total = total + price

        return total

cart = ShoppingCart()
cart.add_item("Shirt", 800)
cart.add_item("Shoes", 1200)
cart.add_item("Bag", 500)

print("Items:", cart.items)
print("Total Bill:", cart.total_bill())
```

30. Create a class Library to store books in a list and add/remove books

```python
class Library:

    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def remove_book(self, book):
        if book in self.books:
            self.books.remove(book)
            print(book, "removed")
        else:
            print(book, "not found")

    def display_books(self):
        print("Books in Library:")

        for book in self.books:
            print(book)


library = Library()
library.add_book("Python Programming")
library.add_book("Java Programming")
library.add_book("Data Science")
library.display_books()
library.remove_book("Java Programming")

print("\nAfter removing:")
library.display_books()
```
